## 1. Definiciones y notacion

Usando el libro **Procesos estocasticos - Luis Rincón (2012)**, una caminata aleatoria simple (o unidimensional 1D) sobre $\mathbb{Z}$ es un proceso estocástico a tiempo discreto $\{X_n : n = 0, 1, \dots\}$ que evoluciona de la siguiente manera:

### Definición de proceso estocastico 1D
Iniciando en el estado $0$ ($X_0 = 0$), en cada unidad de tiempo el proceso pasa al estado $+1$ con probabilidad $p$, o al estado $-1$ con probabilidad $q$, donde se cumple que:
$$p + q = 1$$

Estas probabilidades de transición se escriben formalmente como:
$$P(X_{n+1} = j \mid X_n = i) = \begin{cases} p & \text{si } j = i+1 \\ q & \text{si } j = i-1 \\ 0 & \text{en otro caso} \end{cases}$$

### Construcción mediante variables i.i.d.
El proceso también puede definirse mediante una sucesión $\xi_1, \xi_2, \dots$ de variables aleatorias independientes e idénticamente distribuidas (i.i.d.) tales que:
$$P(\xi = +1) = p \quad \text{y} \quad P(\xi = -1) = q$$

Entonces, para $n \ge 1$, la posición al tiempo $n$ se define como la suma acumulada:
$$X_n := X_0 + \xi_1 + \dots + \xi_n$$



```python
import numpy as np
import matplotlib.pyplot as plt

def simular_caminata_1d(n_pasos, p, n_trayectorias=5):
    q = 1 - p
    pasos_posibles = [1, -1] #izquierda o derecha
    probabilidades = [p, q]

    plt.figure(figsize=(12, 6))

    for i in range(n_trayectorias):
        pasos = np.random.choice(pasos_posibles, size=n_pasos, p=probabilidades)
        trayectoria = np.insert(np.cumsum(pasos), 0, 0)

        plt.plot(trayectoria, label=f'Trayectoria {i+1}')


    plt.title(f'Caminata Aleatoria', fontsize=14)
    plt.xlabel('Pasos (n)', fontsize=12)
    plt.ylabel('Posición (Xn)', fontsize=12)
    plt.axhline(0, color='black', lw=1, ls=':')
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()

# Cambia aqui lo parametros para la caminata aleatoria a tu gusto
N_PASOS = 5000
PROBABILIDAD_P = 0.52
N_TRAYECTORIAS = 8

simular_caminata_1d(N_PASOS, PROBABILIDAD_P, N_TRAYECTORIAS)
```

# Conclusiones

Al observar los resultados de del experimento concluimos  lo siguiente:

1.  **Influencia de ($p$):** La diferencia mínima en la probabilidad (ej. $p=0.51$) hace un desplazamiento importante hacia cualquier dirección a medida que $n$ aumenta.
    

2.  Aunque cada trayectoria individual es impredecible, el promedio de las simulaciones tiende a alinearse con la esperanza $E[X_n] = n(2p-1)$.
